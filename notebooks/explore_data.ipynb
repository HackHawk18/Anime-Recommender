```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= pd.read_csv("E:/programs/Projects/Anime-Recommender/data/anime.csv")
print(df.shape)
print(df.columns)
df.head()
```

```python
print(df.isnull().sum())
print(df['rating'].unique())
print(df.describe())
```

```python

df = df.dropna(subset=['genre'])


df['type'] = df['type'].fillna("Unknown")


df = df.dropna(subset=['rating'])


print(df.shape)
print(df.isnull().sum())
```

```python
print(df['genre'].head(10))

print(df['type'].unique())


print(df['episodes'].unique()[:20])
```

```python

print("Unknown episodes:", (df['episodes'] == 'Unknown').sum())

print("Episodes type:", df['episodes'].dtype)
```

```python
df['episodes'] = df['episodes'].replace('Unknown', 'Ongoing')
print(df['episodes'].value_counts().head(10))
print(df[['name', 'episodes']].head(10))
```

```python

df['name'] = df['name'].str.replace('&#039;', "'", regex=False)
df['name'] = df['name'].str.replace('&amp;', '&', regex=False)

print(df['name'].head(10))
```
